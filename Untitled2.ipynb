```python
import os
import cv2
import numpy as np
from scipy.linalg import eigh
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

def load_images_from_folder(folder):
    images = []
    labels = []
    label_dict = {}
    current_label = 0

    # Assuming the dataset has subfolders for each person
    for person_name in os.listdir(folder):
        person_folder = os.path.join(folder, person_name)
        if not os.path.isdir(person_folder): continue

        label_dict[current_label] = person_name
        for filename in os.listdir(person_folder):
            img_path = os.path.join(person_folder, filename)
            # Read image in grayscale
            img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
            if img is not None:
                # Resize if necessary to ensure all images are the same dimensions
                img = cv2.resize(img, (100, 100))
                # Flatten the image to a column vector (mn x 1)
                images.append(img.flatten())
                labels.append(current_label)
        current_label += 1

    # Transpose so each column is an image: size (mn x p)
    Face_Db = np.array(images).T
    labels = np.array(labels)
    return Face_Db, labels, label_dict
```

```python
Face_Db, labels, label_dict = load_images_from_folder('/content/dataset/faces')
```

```python
import os

folder_path = '/content/dataset'

if os.path.exists(folder_path):
    print(f"Total items in folder: {len(os.listdir(folder_path))}")
    print(f"First few items: {os.listdir(folder_path)[:5]}")
else:
    print(f"ERROR: The folder {folder_path} does not exist!")
```

```python
# Split the data first
X_train, X_test, y_train, y_test = train_test_split(Face_Db.T, labels, test_size=0.40, random_state=42)
Face_Db_train = X_train.T  # Shape: (mn, p_train)
Face_Db_test = X_test.T    # Shape: (mn, p_test)

print(f"Total training images (p): {Face_Db_train.shape[1]}")
print(f"Pixels per image (m*n): {Face_Db_train.shape[0]}")
```

```python
# 2 & 3. Mean Calculation and Mean Zero
M = np.mean(Face_Db_train, axis=1, keepdims=True)
Delta = Face_Db_train - M

# 4. Surrogate Covariance
C = np.dot(Delta.T, Delta)

# 5. Eigen decomposition
eigenvalues, eigenvectors = eigh(C)

# 6. Sort and get Feature vector (Psi) for a default k=50
idx = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[idx]
eigenvectors = eigenvectors[:, idx]

k_default = 50
Psi = eigenvectors[:, :k_default]

# 7. Generating Eigenfaces
Phi = np.dot(Psi.T, Delta.T)

# 8. Generate Signatures
omega = np.dot(Phi, Delta)

print("PCA features and Eigenfaces generated successfully!")
```

```python
import matplotlib.pyplot as plt

k_values = [10, 20, 30, 40, 50, 60, 70]
accuracies = []

# Prepare test data (Mean Zero)
Delta_test = Face_Db_test - M

for current_k in k_values:
    # Extract k directions
    Psi_k = eigenvectors[:, :current_k]
    Phi_k = np.dot(Psi_k.T, Delta.T)

    # Train signatures
    omega_k = np.dot(Phi_k, Delta)

    # Train the ANN
    ann = MLPClassifier(hidden_layer_sizes=(100,), max_iter=1000, random_state=42)
    ann.fit(omega_k.T, y_train)

    # Test the ANN
    Omega_k = np.dot(Phi_k, Delta_test)
    y_pred_k = ann.predict(Omega_k.T)

    # Calculate Accuracy
    acc = accuracy_score(y_test, y_pred_k)
    accuracies.append(acc)
    print(f"Accuracy for k={current_k}: {acc*100:.2f}%")

# Plot the comparative study graph
plt.figure(figsize=(8, 5))
plt.plot(k_values, accuracies, marker='o', linestyle='-', color='b')
plt.title('Classification Accuracy vs. k Value (Eigenvectors)')
plt.xlabel('Number of Eigenvectors (k)')
plt.ylabel('Accuracy')
plt.grid(True)
plt.show()
```

```python
import cv2
import numpy as np

# 1. Load the imposter image
imposter_path = '/content/imposter.jpg'
imposter_img = cv2.imread(imposter_path, cv2.IMREAD_GRAYSCALE)

if imposter_img is not None:
    # Resize to match our database (100x100)
    imposter_img = cv2.resize(imposter_img, (100, 100))

    # Flatten to a column vector (mn x 1)
    imposter_vector = imposter_img.flatten().reshape(-1, 1)

    # 2. Do Mean Zero (Subtract the training mean)
    Delta_imposter = imposter_vector - M

    # 3. Project this mean-aligned face to our eigenfaces to get the signature
    # Note: We use Phi_k from your final loop iteration in Cell 3
    Omega_imposter = np.dot(Phi_k, Delta_imposter)

    # 4. Predict with ANN using probability
    # predict_proba returns the percentage confidence for every single label
    probabilities = ann.predict_proba(Omega_imposter.T)
    max_confidence = np.max(probabilities)

    # Set a threshold (e.g., if confidence is below 50%, it is an imposter)
    THRESHOLD = 0.50

    print("--- IMPOSTER TEST RESULTS ---")
    if max_confidence < THRESHOLD:
        print(f"Result: NOT ENROLLED PERSON (Imposter correctly detected).")
        print(f"Highest Confidence: {max_confidence*100:.2f}% (Below {THRESHOLD*100}% threshold)")
    else:
        # If the ANN is stubbornly confident, see who it thinks the imposter is
        predicted_class = ann.predict(Omega_imposter.T)[0]
        person_name = label_dict.get(predicted_class, f"Person {predicted_class}")
        print(f"Result: Recognized as {person_name}.")
        print(f"Warning: The model was {max_confidence*100:.2f}% confident this imposter was enrolled.")
else:
    print("ERROR: Could not find 'imposter.jpg'. Make sure it is uploaded directly to /content/")
```
